```
#byjus Case
import pandas as pd
import numpy as np

dataset = pd.read_excel("/content/Lecture2_data.xlsx")
```

```
# Decision making structures
# Nested IF Statement in Phython
revenue = dataset["Revenue_Lakh"].mean()
rating = dataset["Customer_Rating"].mean()

if revenue >= 300:
  print("Revenue is high")

  if rating >=4:
    print("Customer satisfaction is also good")
  else:
    print("Customer satisfaction needs improvement")
else:
      print("Revenue is low")
```

```
# Find the average customer rating
rating = dataset["Customer_Rating"].mean()

# Make a decision based on the rating
if rating >= 4.5:
  print("Excellent")
elif rating >= 4.0:
    print("Good")
elif rating >= 3.0:
    print("Average")

else:
        print("Needs Improvement")
```

```
# For Loop
for rating in dataset["Customer_Rating"]:
  if rating >= 4:
    print("Good Rating")
else:
  print("Needs Improvement")
```

```
# For Loop with range() iloc= integer location

for i in range(10):
  print(dataset["Product"].iloc[i], dataset["Revenue_Lakh"].iloc[i])
```

```
#While Loop in Phython
#While loop and i,j is variable to track
i = 0

while i < 5:
  revenue = dataset["Revenue_Lakh"].iloc[i]

  if revenue >= 300:
    print("High Revenue:", revenue)
  else:
   print("Low Revenue:", revenue)

i = i + 1
```
